```python
def load_hri_sense_dialogue(csv_path: Path, source_name: str) -> TranscriptResult:
    segments = []
    with csv_path.open("r", encoding="utf-8", newline="") as handle:
        reader = csv.DictReader(handle)
        for row in reader:
            text = (row.get("text") or "").strip()
            if not text:
                continue
            segments.append(
                TranscriptSegment(
                    start=float(row["start"]),
                    end=float(row["end"]),
                    text=text,
                )
            )

    transcript_text = " ".join(segment.text for segment in segments).strip()
    return TranscriptResult(
        source_name=source_name,
        media_kind="dialogue",
        transcript_text=transcript_text,
        segments=tuple(segments),
        language="en",
        source_path=str(csv_path),
    )


hri_example = load_hri_sense_dialogue(
    Path("../HRI-SENSE/sensory-data/dialogue-transcript/P10-M1-3.csv"),
    "HRI-SENSE example: P10-M1-3",
)
meld_example = VideoEmotionsProcessor().transcribe(
    Path("../MELD.Raw/dev_splits_complete/dia0_utt0.mp4"),
    filename="dia0_utt0.mp4",
)

print("Example dialogue results shown below after the notebook runs.")
display(
    HTML(
        "<div style='margin:0.75rem 0 0.5rem 0;padding:0.75rem 1rem;border-left:4px solid #4c6ef5;background:#f7f9ff;'>"
        "<strong>Notebook examples</strong><br>"
        "HRI-SENSE uses the dialogue transcript CSV. MELD.Raw is transcribed from the spoken video clip."
        "</div>"
    )
)
display(build_results_accordion([hri_example, meld_example]))
```
