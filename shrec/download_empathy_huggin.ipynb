```python
import os
os.environ["HF_HUB_DISABLE_XET"] = "1"

import time
from pathlib import Path
from concurrent.futures import ThreadPoolExecutor, as_completed

import pandas as pd
from dotenv import load_dotenv
from tqdm.auto import tqdm
from huggingface_hub import HfApi, hf_hub_download

load_dotenv()

repo_id = "MIT-personal-robots/shrec_empathic"
repo_type = "dataset"

local_dir = Path.cwd() / "shrec_empathic"
local_dir.mkdir(parents=True, exist_ok=True)

token = os.getenv("accessToken") or True

api = HfApi()

print("Working directory:", Path.cwd())
print("Download folder:", local_dir)
```

```python
def get_remote_files(max_attempts=10):
    for attempt in range(1, max_attempts + 1):
        try:
            print(f"Getting remote file list attempt {attempt}/{max_attempts}...")

            files = api.list_repo_files(
                repo_id=repo_id,
                repo_type=repo_type,
                token=token
            )

            files = sorted(files)

            print(f"Remote total files: {len(files)}")
            print(f"Remote PNG files: {sum(f.lower().endswith('.png') for f in files)}")

            return files

        except Exception as e:
            print(f"Remote file list failed: {e}")
            print("Waiting 30 seconds...")
            time.sleep(30)

    raise RuntimeError("Could not get the remote file list.")


remote_files = get_remote_files()

pd.DataFrame({"remote_file": remote_files}).to_csv(
    "remote_files_shrec_empathic.csv",
    index=False
)

print("Saved remote file list.")
```

```python
def find_missing_files(remote_files):
    missing = []

    for remote_file in remote_files:
        local_path = local_dir / remote_file

        if not local_path.exists():
            missing.append(remote_file)

    return missing


missing_files = find_missing_files(remote_files)

print(f"Missing files before download: {len(missing_files)}")
print(f"Local PNG files before download: {len(list(local_dir.rglob('*.png')))}")
```

```python
def download_one_file(remote_file, max_attempts=8):
    last_error = None

    for attempt in range(1, max_attempts + 1):
        try:
            hf_hub_download(
                repo_id=repo_id,
                repo_type=repo_type,
                filename=remote_file,
                local_dir=str(local_dir),
                token=token,
                force_download=False,
                etag_timeout=90,
                resume_download=True
            )

            local_path = local_dir / remote_file

            if local_path.exists():
                return True, remote_file, None

            last_error = "Download finished, but file was not found locally."

        except Exception as e:
            last_error = str(e)
            wait_time = min(10 * attempt, 90)
            time.sleep(wait_time)

    return False, remote_file, last_error


def download_until_complete(remote_files, workers=2, max_rounds=50):
    for round_number in range(1, max_rounds + 1):
        missing_files = find_missing_files(remote_files)

        remote_png_count = sum(f.lower().endswith(".png") for f in remote_files)
        local_png_count = len(list(local_dir.rglob("*.png")))

        print("\n" + "=" * 70)
        print(f"Download round {round_number}/{max_rounds}")
        print(f"Remote total files: {len(remote_files)}")
        print(f"Remote PNG files: {remote_png_count}")
        print(f"Local PNG files: {local_png_count}")
        print(f"Missing files: {len(missing_files)}")
        print("=" * 70)

        if len(missing_files) == 0:
            print("\nDataset fully downloaded and verified.")
            return True

        pd.DataFrame({"missing_file": missing_files}).to_csv(
            f"missing_files_round_{round_number}.csv",
            index=False
        )

        failed_files = []

        with ThreadPoolExecutor(max_workers=workers) as executor:
            futures = {
                executor.submit(download_one_file, file): file
                for file in missing_files
            }

            progress = tqdm(
                as_completed(futures),
                total=len(futures),
                desc=f"Downloading missing files round {round_number}",
                unit="file"
            )

            completed = 0

            for future in progress:
                success, file_name, error = future.result()
                completed += 1

                percent = (completed / len(missing_files)) * 100
                progress.set_postfix_str(f"{percent:.2f}%")

                if not success:
                    failed_files.append({
                        "file": file_name,
                        "error": error
                    })

        if failed_files:
            pd.DataFrame(failed_files).to_csv(
                f"failed_files_round_{round_number}.csv",
                index=False
            )
            print(f"Failed files this round: {len(failed_files)}")
            print("They will be retried in the next round.")

        print("Waiting 20 seconds before rechecking...")
        time.sleep(20)

    final_missing = find_missing_files(remote_files)

    pd.DataFrame({"missing_file": final_missing}).to_csv(
        "missing_files_final.csv",
        index=False
    )

    raise RuntimeError(
        f"Download incomplete. Missing files left: {len(final_missing)}. "
        "Run the download_until_complete cell again."
    )


download_until_complete(
    remote_files=remote_files,
    workers=2,
    max_rounds=50
)
```

```python
import os
os.environ["HF_HUB_DISABLE_XET"] = "1"

import time
from pathlib import Path
from concurrent.futures import ThreadPoolExecutor, as_completed

import pandas as pd
from dotenv import load_dotenv
from tqdm.auto import tqdm
from huggingface_hub import HfApi, hf_hub_download

load_dotenv()

repo_id = "MIT-personal-robots/shrec_wellness_home"
repo_type = "dataset"

local_dir = Path.cwd() / "shrec_wellness_home"
local_dir.mkdir(parents=True, exist_ok=True)

token = os.getenv("accessToken") or True

api = HfApi()

print("Working directory:", Path.cwd())
print("Download folder:", local_dir)
```

```python
def get_remote_files(max_attempts=10):
    for attempt in range(1, max_attempts + 1):
        try:
            print(f"Getting remote file list attempt {attempt}/{max_attempts}...")

            files = api.list_repo_files(
                repo_id=repo_id,
                repo_type=repo_type,
                token=token
            )

            files = sorted(files)

            print(f"Remote total files: {len(files)}")
            print(f"Remote PNG files: {sum(f.lower().endswith('.png') for f in files)}")

            return files

        except Exception as e:
            print(f"Remote file list failed: {e}")
            print("Waiting 30 seconds...")
            time.sleep(30)

    raise RuntimeError("Could not get the remote file list.")


remote_files = get_remote_files()

pd.DataFrame({"remote_file": remote_files}).to_csv(
    "remote_files_shrec_wellness_home.csv",
    index=False
)

print("Saved remote file list.")
```

```python
def find_missing_files(remote_files):
    missing = []

    for remote_file in remote_files:
        local_path = local_dir / remote_file

        if not local_path.exists():
            missing.append(remote_file)

    return missing


def find_zero_byte_files(remote_files):
    zero_byte_files = []

    for remote_file in remote_files:
        local_path = local_dir / remote_file

        if local_path.exists() and local_path.is_file():
            if local_path.stat().st_size == 0:
                zero_byte_files.append(remote_file)

    return zero_byte_files


def download_one_file(remote_file, max_attempts=8):
    last_error = None

    for attempt in range(1, max_attempts + 1):
        try:
            hf_hub_download(
                repo_id=repo_id,
                repo_type=repo_type,
                filename=remote_file,
                local_dir=str(local_dir),
                token=token,
                force_download=False,
                etag_timeout=90
            )

            local_path = local_dir / remote_file

            if local_path.exists() and local_path.stat().st_size > 0:
                return True, remote_file, None

            last_error = "Download finished, but file was missing or empty."

        except Exception as e:
            last_error = str(e)
            wait_time = min(10 * attempt, 90)
            time.sleep(wait_time)

    return False, remote_file, last_error


def download_until_complete(remote_files, workers=2, max_rounds=50):
    for round_number in range(1, max_rounds + 1):
        missing_files = find_missing_files(remote_files)
        zero_byte_files = find_zero_byte_files(remote_files)

        files_to_download = sorted(set(missing_files + zero_byte_files))

        remote_png_count = sum(f.lower().endswith(".png") for f in remote_files)
        local_png_count = len(list(local_dir.rglob("*.png")))

        print("\n" + "=" * 70)
        print(f"Download round {round_number}/{max_rounds}")
        print(f"Remote total files: {len(remote_files)}")
        print(f"Remote PNG files: {remote_png_count}")
        print(f"Local PNG files: {local_png_count}")
        print(f"Missing files: {len(missing_files)}")
        print(f"Zero-byte files: {len(zero_byte_files)}")
        print(f"Files to download this round: {len(files_to_download)}")
        print("=" * 70)

        if len(files_to_download) == 0:
            print("\nDataset fully downloaded and verified.")
            return True

        pd.DataFrame({"file": files_to_download}).to_csv(
            f"files_to_download_home_round_{round_number}.csv",
            index=False
        )

        failed_files = []

        with ThreadPoolExecutor(max_workers=workers) as executor:
            futures = {
                executor.submit(download_one_file, file): file
                for file in files_to_download
            }

            progress = tqdm(
                as_completed(futures),
                total=len(futures),
                desc=f"Downloading home round {round_number}",
                unit="file"
            )

            completed = 0

            for future in progress:
                success, file_name, error = future.result()
                completed += 1

                percent = (completed / len(files_to_download)) * 100
                progress.set_postfix_str(f"{percent:.2f}%")

                if not success:
                    failed_files.append({
                        "file": file_name,
                        "error": error
                    })

        if failed_files:
            pd.DataFrame(failed_files).to_csv(
                f"failed_files_home_round_{round_number}.csv",
                index=False
            )

            print(f"Failed files this round: {len(failed_files)}")
            print("They will be retried in the next round.")

        print("Waiting 20 seconds before rechecking...")
        time.sleep(20)

    final_missing = find_missing_files(remote_files)

    pd.DataFrame({"missing_file": final_missing}).to_csv(
        "missing_files_home_final.csv",
        index=False
    )

    raise RuntimeError(
        f"Download incomplete. Missing files left: {len(final_missing)}. "
        "Run download_until_complete(remote_files) again."
    )


download_until_complete(
    remote_files=remote_files,
    workers=2,
    max_rounds=50
)
```
